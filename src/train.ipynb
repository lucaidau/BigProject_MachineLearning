```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

x = np.load("../Data/Features/x_features.npy")
y = np.load("../Data/Features/y_labels.npy")

x_train, x_val_test, y_train, y_val_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)

x_val, x_test, y_val, y_test = train_test_split(x_val_test, y_val_test, test_size=0.5, random_state=41, stratify=y_val_test)

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_val_scaled = scaler.transform(x_val)
x_test_scaled = scaler.transform(x_test)

print(f"Train shape: {x_train_scaled.shape}")
print(f"Validation shape: {x_val_scaled.shape}")
print(f"Test shape: {x_test_scaled.shape}")
```
